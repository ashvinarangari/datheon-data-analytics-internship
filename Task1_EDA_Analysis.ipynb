```python
# Sales by Category

category_sales=df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10,6))
sns.barplot(x=category_sales.index, y=category_sales.values)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
# Profit by Category

category_profit=df.groupby("Category")["Profit"].sum().sort_values(ascending= False)

plt.figure(figsize=(10, 6))
sns.barplot(x=category_profit.index, y=category_profit.values)

plt.title("Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
# Monthly Sales Trend

df["Order Date"]= pd.to_datetime(df["Order Date"])

monthly_sales=df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()

plt.figure(figsize=(14, 6))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker='o')

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Sales by Region

region_sales=df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=region_sales.index, y=region_sales.values)

plt.title("Sales by Region", fontsize=16)
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
# Regional sales, profit, and purchase volume
region_summary = df.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Units=("Quantity", "sum")
).sort_values("Sales", ascending=False)

display(region_summary.round(2))
```

```python
# Figures to support the discount findings
discount_summary = df.groupby("Discount").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum"),
    Records=("Sales", "size")
)
discount_summary["Profit Margin %"] = (
    discount_summary["Profit"] / discount_summary["Sales"] * 100
)
display(discount_summary.round(2))
```

## Key Business Findings

### 1. Consumer customers generate the highest sales
The Consumer segment generated sales of 1,161,401.34,
profit of 134,119.21, and purchases totalling 19,521 units.
It contributed more sales than either Corporate or Home Office.

**Business implication:** Prioritise customer service and
retention initiatives for this important customer segment.

### 2. Corporate customers make a substantial contribution
The Corporate segment generated sales of 706,146.37 and
profit of 91,979.13 across 1,514 unique orders.

**Business implication:** Maintain relationships with Corporate
customers through targeted account support and suitable offers.

### 3. Office Supplies leads in quantity sold
Office Supplies accounted for 22,906 units sold, compared
with 8,028 units for Furniture and 6,939 units for Technology.

**Business implication:** Plan inventory around the higher
unit demand for Office Supplies, while also considering
product-level profitability and stock availability.

### 4. Undiscounted transactions have stronger profit margins
Transactions without discounts generated profit of 320,987.60
and an aggregate profit margin of 29.51%. In comparison,
transactions with a 20% discount had a margin of 11.82%.

**Business implication:** Evaluate promotions using profit
margin as well as sales revenue.

### 5. Higher discount groups show aggregate losses
Every observed discount level of 30% or more recorded
negative aggregate profit. For example, the 30% discount
group recorded a loss of 10,369.28.

**Business implication:** Review high-discount transactions
by product and category before continuing similar promotions.
This association does not prove that discounts alone caused
the losses.

## Overall Business Recommendations

- Strengthen retention efforts for Consumer and Corporate customers.
- Align inventory planning with category-level purchase volumes.
- Review the profitability of products receiving high discounts.
- Use monthly sales patterns to support inventory and promotion planning.
- Assess sales and profit together when evaluating business performance.

## Conclusion

The analysis highlights the Consumer segment's leading sales
contribution, the high purchase volume of Office Supplies,
and the losses associated with higher discount groups.

These findings can support customer retention, inventory
planning, and a review of discount policies. The results
describe historical transactions and should be investigated
at product level before making pricing changes.
